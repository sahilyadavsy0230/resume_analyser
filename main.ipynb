```python
from nltk.stem import PorterStemmer,WordNetLemmatizer
from nltk.corpus import stopwords
```

```python
lematizer=WordNetLemmatizer()
```

```python
X=df['Resume']
y=df['Category']
```

```python

import re
def cleanResume(txt):
    cleanText = re.sub(r'http\S+\s', ' ', txt)
    cleanText = re.sub(r'RT|cc', ' ', cleanText)
    cleanText = re.sub(r'#\S+\s', ' ', cleanText)
    cleanText = re.sub(r'@\S+', '  ', cleanText)  
    cleanText = re.sub(r'[%s]' % re.escape(r"""!"#$%&'()*+,-./:;<=>?@[\]^_`{|}~"""), ' ', cleanText)
    cleanText = re.sub(r'[^\x00-\x7f]', ' ', cleanText) 
    cleanText = re.sub(r'\s+', ' ', cleanText)
    cleanText = re.sub(r'[^a-zA-Z]', '  ', cleanText)
    cleanText=cleanText.lower()
    cleanText=cleanText.split()
    cleanText=[lematizer.lemmatize(word) for word in cleanText if word not in stopwords.words('english')]
    cleanText=' '.join(cleanText)
    return cleanText
    
```

```python
df['Resume']=df['Resume'].apply(lambda x:cleanResume(x))
```

```python
df['Resume'][0]
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf=TfidfVectorizer()
```

```python
required_text=tfidf.fit_transform(df['Resume'])
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le=LabelEncoder()
df['Category']=le.fit_transform(df['Category'])
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(required_text,df['Category'],test_size=0.2)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score,confusion_matrix
```

```python
rf=RandomForestClassifier(n_estimators=50)
rf.fit(X_train,Y_train)
Y_pred=rf.predict(X_test)
```

```python
accuracy_score(Y_test,Y_pred)
```
